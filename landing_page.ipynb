# Coherent MXene/metal interfaces: smallest matching supercell and MACE relaxation

Daniel Ocampo

# Abstract

**coherentmxmetal** is a [Sim2L](https://doi.org/10.1371/journal.pone.0264492) [1] that builds a coherent interface
between an MXene (0001) basal plane and the (111) plane of an FCC metal or L1₂ compound, in one orthogonal box,
and can relax it with a MACE foundation model [2].

- **MXene:** ordered MXene (e.g. V₂C, Ti₃C₂, Mo₂TiC₂O₂) in its a × √3a cell, with AβC (octahedral) or AβA
  (trigonal prismatic) stacking, optionally turned 180° about z.
- **Metal:** close-packed (111) layers stacked A → B → C (or its twin), with the first layer on a chosen site of
  the MXene grid at a set gap above the MXene.
- **Supercell:** the smallest whole numbers of MXene and metal replicas whose lengths match within the strain
  tolerance; the metal takes the (equi-biaxial) strain.
- **Relaxation (optional):** MACE-MH-1 in float64 with FIRE, outermost metal layers fixed, in-plane cell free,
  and the work of adhesion.

Lattice constants come from the bundled spreadsheet (41 MXenes, 23 metals). Structures are built with
pymatgen [3] and relaxed with ASE [4].

# Get started

- [**Workflow notebook**](bin/workflow.ipynb): set the inputs, run the Sim2L, and look at the supercell, strain,
  structure and (after a relaxation) energy, gap and work of adhesion. The POSCARs can be downloaded.
- [**Sim2L notebook**](simtool/coherentmxmetal.ipynb): the declared inputs and outputs and the workflow itself.
- Every run is cached and indexed in nanoHUB's ResultsDB, so runs with the same inputs return immediately.
- Source code, a web app and command-line versions:
  [github.com/docampomill/coherent-interface-builder](https://github.com/docampomill/coherent-interface-builder)
